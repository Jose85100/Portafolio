# Demanda de crédito: endogeneidad y variables instrumentales

**Laboratorio crítico | Datos sintéticos | No evidencia empírica**

La tasa se fija según información del prestamista; una elasticidad OLS puede reflejar pricing por riesgo y no demanda. El experimento enfrenta OLS con IV bajo exclusión controlada.

## 1. Hipótesis rivales y supuestos
**DGP:** y=βx+u, x=πz+δu+v, donde z,u,v son independientes normales estándar. β=−1.5 es la elasticidad estructural del ejemplo. z es válido solo en el DGP construido; un instrumento real requiere justificación.

## 2. Derivación
plim β_OLS = β+δ/(π²+δ²+1). plim β_IV=Cov(z,y)/Cov(z,x)=β si π≠0 y Cov(z,u)=0. Cuando π es pequeño, el IV puede ser muy inestable en muestras finitas.


```python
# Preparación reproducible y funciones del experimento
import random, math
import matplotlib.pyplot as plt
def slopes(pi,n=3000,seed=18,beta=-1.5,delta=.8):
    rng=random.Random(seed);rows=[]
    for _ in range(n):
        z,u,v=[rng.gauss(0,1) for j in range(3)]
        x=pi*z+delta*u+v;y=beta*x+u
        rows.append((x,y,z))
    def cov(a,b):
        ma=sum(a)/len(a);mb=sum(b)/len(b)
        return sum((x-ma)*(y-mb) for x,y in zip(a,b))/len(a)
    x,y,z=zip(*rows)
    return cov(x,y)/cov(x,x),cov(z,y)/cov(z,x)

```

## 3. Experimento: escenario base y ruptura de supuestos

La comparación mantiene constantes los demás parámetros. Un cambio de resultado al violar un supuesto **no** invalida una proposición que lo requiere.


```python
strength=[.2,.4,.7,1,1.5,2]
ols,iv=zip(*(slopes(p) for p in strength))
plt.figure(figsize=(9,4.5))
plt.plot(strength,ols,'o-',label='OLS')
plt.plot(strength,iv,'s-',label='IV')
plt.axhline(-1.5,color='gray',linestyle='--',label='Elasticidad estructural')
plt.xlabel('Relevancia instrumental π');plt.ylabel('Coeficiente estimado')
plt.title('Precio endógeno: sesgo OLS e inestabilidad IV')
plt.legend();plt.grid(alpha=.2);plt.show()
assert all(math.isfinite(v) for v in ols+iv)
```

## 4. Interpretación y falsación

El precio puede alterar aprobación, composición de solicitantes y demanda extensiva. Un shock de fondeo puede violar exclusión si cambia simultáneamente límites de crédito. Para validar IV real, examinar primera etapa, intervalos robustos a instrumentos débiles y mecanismos directos. No interpretar desembolsos como demanda solicitada.
