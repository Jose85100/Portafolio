# Mixed regret under action-specific factor loadings

This is a deterministic synthetic 2x2 sensitivity exercise. It estimates no empirical behavior.

## Question and exact bound
For `u[i,j]=A[i,j]+lambda[i]*f[j]`, with `|f[j]|<=delta`, the regret bound is `max(0, max_a(g[a] + delta*abs(lambda[a]-p.lambda)*sum(q)))`. Equal loadings cancel; heterogeneous loadings leave a support term.

## Configured comparison
The notebook displays every configured sensitivity row and the analytic threshold crossing. `independent_cell_rectangular` is an adversarial ablation of the assumed error structure.

```python
import csv, json
from pathlib import Path
rows = list(csv.DictReader((Path.cwd() / 'results/comparison.csv').open()))
print('delta | structure | regret | threshold_exceeded')
print('---:|:---|---:|:---')
for row in rows:
    print(f"{row['delta']} | {row['structure']} | {float(row['regret']):.3f} | {row['threshold_exceeded']}")
summary = json.loads((Path.cwd() / 'results/summary.json').read_text())
print('\nSummary:\n' + json.dumps(summary, indent=2))
```
