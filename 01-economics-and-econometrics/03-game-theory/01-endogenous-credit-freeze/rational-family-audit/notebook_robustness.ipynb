# ¿Cuánto error en pagos elimina el certificado de equilibrio?

## Mecanismo y fórmula
Para el juego de coordinación con pagos $A=[[1,-1],[-1,1]]$ y mezcla $(1/2,1/2)$, cada jugador es indiferente entre acciones y regret nominal es cero. Si cada pago puede variar independientemente en $[-\delta,\delta]$, el peor regret contra una desviación pura es $\max(0,g_a+\delta \sum_{ij}q_j|\mathbf1(i=a)-p_i|)$. La expresión añadida es el soporte exacto de la caja de errores; no es una cota probabilística ni un nuevo equilibrio.

El notebook usa la API tipada del programa y compara con el oráculo exhaustivo de las 16 esquinas. El único escenario es sintético y deliberado.

```python
from src.epsilon_robustness import robust_regret
A=((1.,-1.),(-1.,1.)); p=q=(.5,.5)
vals=[(d, robust_regret(A,p,q,d).nominal_regret, robust_regret(A,p,q,d).robust_regret) for d in (0,.005,.01,.02,.05,.1)]
assert all(abs(r-d)<1e-12 for d,_,r in vals)
vals
```

## Lectura crítica
La línea nominal es cero, mientras que el máximo sobre la caja crece con $\delta$. El certificado exacto de Nash del juego sin ruido no es estable frente a error adversarial de pagos. Esto no estima la distribución real del error. Si los errores están correlacionados, el set admisible debe cambiar y el resultado no aplica. La extensión clasifica como auditoría/adaptación de ingeniería sobre la línea Endogenous Credit Freeze; no reclama nueva teoría de equilibrio.
