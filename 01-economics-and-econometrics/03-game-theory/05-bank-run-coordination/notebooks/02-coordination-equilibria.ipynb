# Corridas bancarias: coordinación y multiplicidad de equilibrio

**Laboratorio crítico | Datos sintéticos | No evidencia empírica**

Un juego de dos depositantes puede sostener tanto retirar como mantener fondos, dependiendo de expectativas. Contrastamos dominancia, coordinación y selección de equilibrio.

## 1. Hipótesis rivales y supuestos
Cada jugador elige retirar (R) o mantener (M). Pagos simétricos: (M,M)=(3,3), (R,R)=(1,1), (R,M)=(2,0), (M,R)=(0,2). Dos equilibrios puros: (M,M) y (R,R). Es un juego ilustrativo, no un modelo calibrado de bancos.

## 2. Derivación
Si el rival retira con probabilidad p, E[U(R)]=2(1−p)+p=2−p; E[U(M)]=3(1−p)+0p=3−3p. Indiferencia en p=1/2. El juego tiene dos Nash puros y uno mixto con p=0.5. Modificar los pagos puede eliminar multiplicidad.


```python
# Preparación reproducible y funciones del experimento
import matplotlib.pyplot as plt
payoffs={('M','M'):(3,3),('R','R'):(1,1),('R','M'):(2,0),('M','R'):(0,2)}
def best_responses():
    eq=[]
    for a in ('M','R'):
        for b in ('M','R'):
            ua,ub=payoffs[(a,b)]
            if ua>=payoffs[('M' if a=='R' else 'R',b)][0] and ub>=payoffs[(a,'M' if b=='R' else 'R')][1]:eq.append((a,b))
    return eq
assert set(best_responses())=={('M','M'),('R','R')}

```

## 3. Experimento: escenario base y ruptura de supuestos

La comparación mantiene constantes los demás parámetros. Un cambio de resultado al violar un supuesto **no** invalida una proposición que lo requiere.


```python
p=[i/100 for i in range(101)]
ur=[2-x for x in p];um=[3-3*x for x in p]
assert abs(ur[50]-um[50])<1e-12
plt.figure(figsize=(9,4.5))
plt.plot(p,ur,label='Retirar')
plt.plot(p,um,label='Mantener')
plt.axvline(.5,color='gray',linestyle='--',label='Equilibrio mixto')
plt.xlabel('Creencia P(rival retira)');plt.ylabel('Pago esperado')
plt.title('Corrida bancaria: mejores respuestas y coordinación')
plt.legend();plt.grid(alpha=.2);plt.show()
print('Equilibrios puros',best_responses())
```

## 4. Interpretación y falsación

La multiplicidad no prueba que toda corrida sea autorrealizada. Liquidez, solvencia, seguro de depósitos, información privada y garantías modifican los pagos y pueden seleccionar equilibrios.
