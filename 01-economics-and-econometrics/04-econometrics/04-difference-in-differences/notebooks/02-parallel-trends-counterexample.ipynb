# DiD: falsación de tendencias paralelas

**Laboratorio crítico | Datos sintéticos | No evidencia empírica**

Un efecto de tratamiento positivo puede sobreestimarse cuando los tratados habrían crecido más incluso sin intervención. Distinguimos ATT del contraste observado.

## 1. Hipótesis rivales y supuestos
**Caso base:** tendencias contrafactuales iguales. **Caso rival:** tendencia diferencial δ. El efecto real ATT=2. No se infiere paralelismo solo porque un placebo pretratamiento sea no significativo.

## 2. Derivación
DiD = (Y_T,post−Y_T,pre)−(Y_C,post−Y_C,pre). Bajo tendencia diferencial δ, E[DiD]=ATT+δ. Con δ=1.5, el límite es 3.5, no 2.


```python
# Preparación reproducible y funciones del experimento
import random
import matplotlib.pyplot as plt
def did_sim(delta,n=4000,seed=7):
    rng=random.Random(seed)
    obs=[]
    for _ in range(n):
        b=rng.gauss(0,1)
        tp=2+b+rng.gauss(0,.5)
        tpost=3+b+2+delta+rng.gauss(0,.5)
        cp=b+rng.gauss(0,.5)
        cpost=1+b+rng.gauss(0,.5)
        obs.append((tp,tpost,cp,cpost))
    means=[sum(row[j] for row in obs)/n for j in range(4)]
    return (means[1]-means[0])-(means[3]-means[2])

```

## 3. Experimento: escenario base y ruptura de supuestos

La comparación mantiene constantes los demás parámetros. Un cambio de resultado al violar un supuesto **no** invalida una proposición que lo requiere.


```python
deltas=[-1,-.5,0,.5,1,1.5,2]
estimates=[did_sim(d) for d in deltas]
assert max(abs(est-(2+d)) for est,d in zip(estimates,deltas))<.08
plt.figure(figsize=(9,4.5))
plt.plot(deltas,estimates,'o-',label='DiD sintético')
plt.plot(deltas,[2+d for d in deltas],'--',label='ATT + sesgo')
plt.axhline(2,color='gray',linestyle=':',label='ATT verdadero')
plt.xlabel('Diferencia de tendencias δ');plt.ylabel('Efecto estimado')
plt.title('Identificación DiD y ruptura de tendencias paralelas')
plt.legend();plt.grid(alpha=.2);plt.show()
```

## 4. Interpretación y falsación

Examinar composición de grupos, shocks simultáneos, anticipación y spillovers. Comparar control sintético y estimadores por cohortes bajo adopción escalonada. Un pretrend estadísticamente insignificante puede reflejar baja potencia; la condición contrafactual post no es directamente verificable.
