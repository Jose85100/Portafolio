# Curva de Phillips: identificación y cambio de signo

**Laboratorio crítico | Datos sintéticos | No evidencia empírica**

¿Puede una pendiente reducida negativa coexistir con una relación estructural positiva? Comparamos un sistema sin reacción endógena y otro con retroalimentación de política.

## 1. Hipótesis rivales y supuestos
**Modelo estructural simplificado:** π = κx + u, x = v − au. Se suponen u,v normales estándar independientes y expectativas constantes. No es una NKPC completa. **Alternativas:** expectativas adaptativas, sticky information, shocks de oferta y no linealidad.

## 2. Derivación
Como Cov(x,u)=−a y Var(x)=1+a², plim β_OLS = κ−a/(1+a²). Para κ=0.3 y a=1, la pendiente observacional converge a −0.2. La inversión de signo es un problema de identificación, no prueba de κ negativa.


```python
# Preparación reproducible y funciones del experimento
import random, math
import matplotlib.pyplot as plt
rng=random.Random(19)
def experiment(a,n=5000,kappa=.3):
    pairs=[]
    for _ in range(n):
        u,v=rng.gauss(0,1),rng.gauss(0,1)
        x=v-a*u
        pairs.append((x,kappa*x+u))
    mx=sum(x for x,y in pairs)/n;my=sum(y for x,y in pairs)/n
    slope=sum((x-mx)*(y-my) for x,y in pairs)/sum((x-mx)**2 for x,y in pairs)
    return slope

```

## 3. Experimento: escenario base y ruptura de supuestos

La comparación mantiene constantes los demás parámetros. Un cambio de resultado al violar un supuesto **no** invalida una proposición que lo requiere.


```python
feedback=[0,.25,.5,.75,1,1.5,2]
observed=[experiment(a) for a in feedback]
expected=[.3-a/(1+a*a) for a in feedback]
assert max(abs(x-y) for x,y in zip(observed,expected))<.07
plt.figure(figsize=(9,4.5))
plt.plot(feedback,observed,'o-',label='OLS sintético')
plt.plot(feedback,expected,'--',label='Límite teórico OLS')
plt.axhline(.3,color='gray',linestyle=':',label='Pendiente estructural')
plt.xlabel('Retroalimentación a');plt.ylabel('Pendiente inflación/brecha')
plt.title('Endogeneidad: el signo observado no identifica κ')
plt.legend();plt.grid(alpha=.2);plt.show()
print(list(zip(feedback,observed)))
```

## 4. Interpretación y falsación

El supuesto de independencia de shocks permite la derivación. Si la política responde a inflación no observada, OLS confunde respuesta de equilibrio y pendiente estructural. Contrastar con expectativas observadas e instrumentos cuya exclusión pueda defenderse; analizar inflación importada y regímenes. No atribuir cambios en correlación a cambios en κ sin evidencia.
