# Incertidumbre de LGD con riesgos competidores

Pregunta: ¿qué componente de varianza se pierde al sustituir LGD incierta por su media?

Con hazards constantes, la densidad incondicional de default es $\lambda_d e^{-(\lambda_d+\lambda_p)t}$. Si $EAD(t)=a+bt$ y $G\sim Beta(\alpha,\beta)$ independiente, $E[L]=E[G]\lambda_d\int_0^T e^{-ht}(a+bt)dt$ y $E[L^2]=E[G^2]\lambda_d\int_0^T e^{-ht}(a+bt)^2dt$. La diferencia entre momentos completos y plug-in cuantifica únicamente dispersión LGD omitida; no es predicción empírica.

La línea previa ya integraba EAD×LGD afines deterministas; este incremento añade parámetro LGD aleatorio independiente para aislar incertidumbre. Se compara con baseline plug-in de igual media.

```python
from src.lgd_uncertainty import Scenario, evaluate
s = Scenario(100.0,160.0,1.0,.18,.12,3.0,2.0)
r = evaluate(s)
print(f"Default probability: {r.default_probability:.12f}")
print(f"Expected loss: {r.expected_loss:.12f}")
print(f"SD uncertain LGD: {r.loss_sd:.12f}")
print(f"SD plug-in mean LGD: {r.plugin_variance**0.5:.12f}")
assert r.loss_variance >= r.plugin_variance >= 0
```
