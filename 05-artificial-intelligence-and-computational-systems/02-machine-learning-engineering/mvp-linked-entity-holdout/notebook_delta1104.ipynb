# Reproducibilidad de vínculos bajo deltas tardíos
Distinguimos tiempo válido $v$ de disponibilidad $a$. Para identidad consultada en $t$ y corte $c$, la historia admisible es $H(t,c)=\{e:a_e\le c, v_e\le t\}$. El estado por arista toma la acción de mayor $v$; llegada tardía no reescribe el orden causal.

```python
from src.delta_replay import Delta, replay, snapshot_first
events=[Delta('a|b','link','2026-01-01T00:00:00Z','2026-01-01T00:00:00Z','e1'), Delta('a|b','unlink','2026-01-02T00:00:00Z','2026-01-02T00:00:00Z','e2'), Delta('a|b','link','2026-01-01T12:00:00Z','2026-01-03T00:00:00Z','e3')]
q={'identity_at':'2026-01-03T00:00:00Z','cutoff':'2026-01-05T00:00:00Z'}
assert replay(events,**q)==set()
assert snapshot_first(events,**q)=={'a|b'}
print('oraculo={} baseline={}'.format(replay(events,**q),snapshot_first(events,**q)))
```

La tercera entrada llega después, pero modifica una vigencia anterior al `unlink`. El baseline de última llegada re-activa erróneamente el vínculo. Esta extensión no se extrapola a datos autenticados ni a deltas con otra semántica.
